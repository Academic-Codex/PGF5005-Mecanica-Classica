### Exercício 2.1 (Teórico)

**Demonstre que a Hamiltoniana H é uma constante de movimento para quaisquer valores dos parâmetros $\alpha$ e $\beta$. Em seguida, mostre que as funções $F_1 = J_1 + J_2$, para $\beta = 0$, e $F_2 = 3J_1 + 2J_2$, para $\alpha = 0$, também representam constantes de movimento.**

---

#### a) H como Constante de Movimento

A evolução temporal de uma função $F(q, p, t)$ na mecânica Hamiltoniana é dada por:
$$ 
\frac{dF}{dt} = \{F, H\} + \frac{\partial F}{\partial t}
$$
onde $\{F, H\}$ é o Parêntese de Poisson. Para a Hamiltoniana $H$ em si, temos:
$$ 
\frac{dH}{dt} = \{H, H\} + \frac{\partial H}{\partial t}
$$
A Hamiltoniana de Walker-Ford, $H = H_0 + H_1$, não depende explicitamente do tempo, logo:
$$ 
\frac{\partial H}{\partial t} = 0
$$
O Parêntese de Poisson de qualquer função com ela mesma é sempre nulo:
$$ 
\{H, H\} = \sum_{k=1}^{2} \left( \frac{\partial H}{\partial \phi_k} \frac{\partial H}{\partial J_k} - \frac{\partial H}{\partial J_k} \frac{\partial H}{\partial \phi_k} \right) = 0
$$
Portanto, para quaisquer valores de $\alpha$ e $\beta$:
$$ 
\frac{dH}{dt} = 0 + 0 = 0
$$
Isso demonstra que a energia total, representada pela Hamiltoniana H, é uma constante de movimento.

#### b) $F_1 = J_1 + J_2$ como Constante de Movimento para $\beta = 0$

Quando $\beta=0$, a Hamiltoniana se torna:
$$ 
H|_{\beta=0} = J_1 + J_2 - J_1^2 - 3J_1J_2 + J_2^2 + \alpha J_1J_2 \cos(2\phi_1 - 2\phi_2)
$$
A função $F_1 = J_1 + J_2$ não depende explicitamente do tempo, então $\frac{\partial F_1}{\partial t} = 0$. Precisamos calcular o Parêntese de Poisson $\{F_1, H\}$:
$$ 
\{F_1, H\} = \frac{\partial F_1}{\partial \phi_1} \frac{\partial H}{\partial J_1} - \frac{\partial F_1}{\partial J_1} \frac{\partial H}{\partial \phi_1} + \frac{\partial F_1}{\partial \phi_2} \frac{\partial H}{\partial J_2} - \frac{\partial F_1}{\partial J_2} \frac{\partial H}{\partial \phi_2}
$$
As derivadas de $F_1$ são $\frac{\partial F_1}{\partial J_1} = 1$, $\frac{\partial F_1}{\partial J_2} = 1$ e $\frac{\partial F_1}{\partial \phi_k} = 0$. Substituindo:
$$ 
\{F_1, H\} = 0 - (1)\frac{\partial H}{\partial \phi_1} + 0 - (1)\frac{\partial H}{\partial \phi_2} = -\left( \frac{\partial H}{\partial \phi_1} + \frac{\partial H}{\partial \phi_2} \right)
$$
Calculando as derivadas de H:
$$ 
\frac{\partial H}{\partial \phi_1} = -\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2) \cdot (2) = -2\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2)
$$
$$ 
\frac{\partial H}{\partial \phi_2} = -\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2) \cdot (-2) = 2\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2)
$$
Somando as duas:
$$ 
\frac{\partial H}{\partial \phi_1} + \frac{\partial H}{\partial \phi_2} = -2\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2) + 2\alpha J_1 J_2 \sin(2\phi_1 - 2\phi_2) = 0
$$
Portanto, $\{F_1, H\} = 0$, o que implica $\frac{dF_1}{dt} = 0$. $F_1$ é uma constante de movimento quando $\beta = 0$.

#### c) $F_2 = 3J_1 + 2J_2$ como Constante de Movimento para $\alpha = 0$

Quando $\alpha=0$, a Hamiltoniana é:
$$ 
H|_{\alpha=0} = J_1 + J_2 - J_1^2 - 3J_1J_2 + J_2^2 + \beta J_1J_2^{3/2} \cos(2\phi_1 - 3\phi_2)
$$
Para $F_2 = 3J_1 + 2J_2$, temos $\frac{\partial F_2}{\partial J_1} = 3$, $\frac{\partial F_2}{\partial J_2} = 2$ e $\frac{\partial F_2}{\partial \phi_k} = 0$. O Parêntese de Poisson é:
$$ 
\{F_2, H\} = -\frac{\partial F_2}{\partial J_1} \frac{\partial H}{\partial \phi_1} - \frac{\partial F_2}{\partial J_2} \frac{\partial H}{\partial \phi_2} = -\left( 3\frac{\partial H}{\partial \phi_1} + 2\frac{\partial H}{\partial \phi_2} \right)
$$
Calculando as derivadas de H:
$$ 
\frac{\partial H}{\partial \phi_1} = -\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2) \cdot (2) = -2\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2)
$$
$$ 
\frac{\partial H}{\partial \phi_2} = -\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2) \cdot (-3) = 3\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2)
$$
Substituindo no Parêntese de Poisson:
$$ 
3\frac{\partial H}{\partial \phi_1} + 2\frac{\partial H}{\partial \phi_2} = 3(-2\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2)) + 2(3\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2))
$$ 
$$ 
= -6\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2) + 6\beta J_1 J_2^{3/2} \sin(2\phi_1 - 3\phi_2) = 0
$$
Logo, $\{F_2, H\} = 0$, e $\frac{dF_2}{dt} = 0$. $F_2$ é uma constante de movimento quando $\alpha = 0$.

#### Conclusão
Um sistema Hamiltoniano com N graus de liberdade é dito integrável se possuir N constantes de movimento independentes e em involução. O sistema de Walker-Ford tem N=2 graus de liberdade. Em ambos os casos, $\alpha=0$ ou $\beta=0$, o sistema possui duas constantes de movimento independentes (H e $F_1$, ou H e $F_2$). Isso confirma que o sistema é integrável quando uma das perturbações ressonantes é nula.

### Exercício 2.2 (Teórico)
**Considerando a Hamiltoniana (22a), escreva as equações para $\phi_{k}^{(n+1)}$ e $J_{k}^{(n+1)}$ de acordo com o método de Euler simplético.**

---

As equações de Hamilton para o sistema são:
$$ 
\dot{\phi}_k = \frac{\partial H}{\partial J_k} \quad , \quad \dot{J}_k = -\frac{\partial H}{\partial \phi_k} 
$$
O método de Euler simplético, conforme as equações (4a) e (4b) do enunciado, com a identificação $q \rightarrow \phi$ e $p \rightarrow J$, é dado por:
$$ 
\phi_{k}^{(n+1)} = \phi_{k}^{(n)} + \Delta t \frac{\partial H}{\partial J_{k}}\bigg|_{(\phi^{(n+1)}, J^{(n)})} \quad \text{(Implícito)} 
$$ 
$$ 
J_{k}^{(n+1)} = J_{k}^{(n)} - \Delta t \frac{\partial H}{\partial \phi_{k}}\bigg|_{(\phi^{(n+1)}, J^{(n)})} \quad \text{(Explícito, após encontrar } \phi^{(n+1)})
$$
Primeiro, calculamos as derivadas parciais da Hamiltoniana completa:
$$ 
H = J_1 + J_2 - J_1^2 - 3J_1J_2 + J_2^2 + \alpha J_1J_2\cos(2\phi_1 - 2\phi_2) + \beta J_1J_2^{\frac{3}{2}}\cos(2\phi_1 - 3\phi_2)
$$
Derivadas em relação a $J_k$:
$$ 
\frac{\partial H}{\partial J_1} = 1 - 2J_1 - 3J_2 + \alpha J_2\cos(2\phi_1 - 2\phi_2) + \beta J_2^{\frac{3}{2}}\cos(2\phi_1 - 3\phi_2)
$$
$$ 
\frac{\partial H}{\partial J_2} = 1 - 3J_1 + 2J_2 + \alpha J_1\cos(2\phi_1 - 2\phi_2) + \frac{3}{2}\beta J_1J_2^{\frac{1}{2}}\cos(2\phi_1 - 3\phi_2)
$$
Derivadas em relação a $\phi_k$:
$$ 
\frac{\partial H}{\partial \phi_1} = -2\alpha J_1J_2\sin(2\phi_1 - 2\phi_2) - 2\beta J_1J_2^{\frac{3}{2}}\sin(2\phi_1 - 3\phi_2)
$$
$$ 
\frac{\partial H}{\partial \phi_2} = 2\alpha J_1J_2\sin(2\phi_1 - 2\phi_2) + 3\beta J_1J_2^{\frac{3}{2}}\sin(2\phi_1 - 3\phi_2)
$$
Agora, substituímos essas derivadas nas equações do método de Euler simplético.

**Equações para $\phi_k^{(n+1)}$ (Sistema Implícito):**
Para $k=1$:
$$ 
\phi_{1}^{(n+1)} = \phi_{1}^{(n)} + \Delta t \left[ 1 - 2J_1^{(n)} - 3J_2^{(n)} + \alpha J_2^{(n)}\cos(2\phi_1^{(n+1)} - 2\phi_2^{(n+1)}) + \beta (J_2^{(n)})^{\frac{3}{2}}\cos(2\phi_1^{(n+1)} - 3\phi_2^{(n+1)}) \right]
$$
Para $k=2$:
$$ 
\phi_{2}^{(n+1)} = \phi_{2}^{(n)} + \Delta t \left[ 1 - 3J_1^{(n)} + 2J_2^{(n)} + \alpha J_1^{(n)}\cos(2\phi_1^{(n+1)} - 2\phi_2^{(n+1)}) + \frac{3}{2}\beta J_1^{(n)}(J_2^{(n)})^{\frac{1}{2}}\cos(2\phi_1^{(n+1)} - 3\phi_2^{(n+1)}) \right]
$$
Este é um sistema de duas equações não-lineares para as duas incógnitas $\phi_1^{(n+1)}$ e $\phi_2^{(n+1)}$, que precisa ser resolvido numericamente a cada passo de tempo (e.g., pelo método de Newton).

**Equações para $J_k^{(n+1)}$ (Sistema Explícito):**
Uma vez encontrados os valores de $\phi_1^{(n+1)}$ e $\phi_2^{(n+1)}$, podemos calcular diretamente $J_k^{(n+1)}$:
Para $k=1$:
$$ 
J_{1}^{(n+1)} = J_{1}^{(n)} - \Delta t \left[ -2\alpha J_1^{(n)}J_2^{(n)}\sin(2\phi_1^{(n+1)} - 2\phi_2^{(n+1)}) - 2\beta J_1^{(n)}(J_2^{(n)})^{\frac{3}{2}}\sin(2\phi_1^{(n+1)} - 3\phi_2^{(n+1)}) \right]
$$
Para $k=2$:
$$ 
J_{2}^{(n+1)} = J_{2}^{(n)} - \Delta t \left[ 2\alpha J_1^{(n)}J_2^{(n)}\sin(2\phi_1^{(n+1)} - 2\phi_2^{(n+1)}) + 3\beta J_1^{(n)}(J_2^{(n)})^{\frac{3}{2}}\sin(2\phi_1^{(n+1)} - 3\phi_2^{(n+1)}) \right]
$$

### Hamiltoniana do sistema e derivadas

```python
# wf/hamiltonian.py
from dataclasses import dataclass
import numpy as np

@dataclass(frozen=True)
class WFParams:
    alpha: float
    beta: float

def H(phi1, phi2, J1, J2, p: WFParams):
    H0 = J1 + J2 - J1**2 - 3*J1*J2 + J2**2
    term_a = p.alpha * J1 * J2 * np.cos(2*phi1 - 2*phi2)
    term_b = p.beta  * J1 * J2**1.5 * np.cos(2*phi1 - 3*phi2)
    val = H0 + term_a + term_b
    return np.where((J1<0)|(J2<0), np.inf, val)

def dH_dJ1(phi1,phi2,J1,J2,p):
    return 1 - 2*J1 - 3*J2 + p.alpha*J2*np.cos(2*phi1-2*phi2) + p.beta*J2**1.5*np.cos(2*phi1-3*phi2)

def dH_dJ2(phi1,phi2,J1,J2,p):
    return 1 - 3*J1 + 2*J2 + p.alpha*J1*np.cos(2*phi1-2*phi2) + 1.5*p.beta*J1*J2**0.5*np.cos(2*phi1-3*phi2)

def dH_dphi1(phi1,phi2,J1,J2,p):
    return -2*p.alpha*J1*J2*np.sin(2*phi1-2*phi2) - 2*p.beta*J1*J2**1.5*np.sin(2*phi1-3*phi2)

def dH_dphi2(phi1,phi2,J1,J2,p):
    return  2*p.alpha*J1*J2*np.sin(2*phi1-2*phi2) + 3*p.beta*J1*J2**1.5*np.sin(2*phi1-3*phi2)

# segundas derivadas (para o Jacobiano do Newton)
def d2H_dJ1dphi1(phi1,phi2,J1,J2,p):
    return -2*p.alpha*J2*np.sin(2*phi1-2*phi2) - 2*p.beta*J2**1.5*np.sin(2*phi1-3*phi2)

def d2H_dJ1dphi2(phi1,phi2,J1,J2,p):
    return  2*p.alpha*J2*np.sin(2*phi1-2*phi2) + 3*p.beta*J2**1.5*np.sin(2*phi1-3*phi2)

def d2H_dJ2dphi1(phi1,phi2,J1,J2,p):
    return -2*p.alpha*J1*np.sin(2*phi1-2*phi2) - 3*p.beta*J1*J2**0.5*np.sin(2*phi1-3*phi2)

def d2H_dJ2dphi2(phi1,phi2,J1,J2,p):
    return  2*p.alpha*J1*np.sin(2*phi1-2*phi2) + 4.5*p.beta*J1*J2**0.5*np.sin(2*phi1-3*phi2)
```

### LU Solver

```python
# wf/lu.py
import numpy as np

def lu_decomp_pivot(A):
    A = A.copy().astype(float)
    n = A.shape[0]
    P = np.eye(n); L = np.zeros_like(A); U = A.copy()
    for k in range(n):
        i = np.argmax(np.abs(U[k:,k])) + k
        if i != k:
            U[[k,i]] = U[[i,k]]
            P[[k,i]] = P[[i,k]]
            L[[k,i],:k] = L[[i,k],:k]
        L[k,k] = 1.0
        for j in range(k+1,n):
            L[j,k] = U[j,k]/U[k,k]
            U[j,k:] -= L[j,k]*U[k,k:]
    return P,L,U

def lu_solve(P,L,U,b):
    Pb = P @ b
    # Ly = Pb
    y = np.zeros_like(b, dtype=float)
    for i in range(L.shape[0]):
        y[i] = Pb[i] - L[i,:i] @ y[:i]
    # Ux = y
    x = np.zeros_like(b, dtype=float)
    for i in range(U.shape[0]-1, -1, -1):
        x[i] = (y[i] - U[i,i+1:] @ x[i+1:]) / U[i,i]
    return x
```

### Método de Newton

```python
# wf/newton.py
import numpy as np

def newton(g, J, x0, lin_solve, tol=1e-10, max_iter=20):
    x = np.array(x0, dtype=float)
    for _ in range(max_iter):
        r = g(x)
        if np.linalg.norm(r, ord=2) < tol:
            return x
        A = J(x)
        dx = lin_solve(A, -r)
        x = x + dx
    return x  # melhor aproximação
```

### Euler Simplético Implícito

```python
# wf/symplectic_euler.py
import numpy as np


def step_symplectic_euler(state, dt, params, newton_solve):
    phi = state[:2].copy(); J = state[2:].copy()
    J1,J2 = J

    def g(x):
        φ1,φ2 = x
        return np.array([
            φ1 - phi[0] - dt * dH_dJ1(φ1,φ2,J1,J2,params),
            φ2 - phi[1] - dt * dH_dJ2(φ1,φ2,J1,J2,params),
        ])

    def Jg(x):
        φ1,φ2 = x
        return np.array([
            [1 - dt*d2H_dJ1dphi1(φ1,φ2,J1,J2,params), -dt*d2H_dJ1dphi2(φ1,φ2,J1,J2,params)],
            [-dt*d2H_dJ2dphi1(φ1,φ2,J1,J2,params),      1 - dt*d2H_dJ2dphi2(φ1,φ2,J1,J2,params)]
        ])

    phi_next = newton_solve(g, Jg, phi)   # resolve o passo implícito
    φ1,φ2 = phi_next
    J_next = np.array([
        J1 - dt * dH_dphi1(φ1,φ2,J1,J2,params),
        J2 - dt * dH_dphi2(φ1,φ2,J1,J2,params),
    ])
    return np.concatenate([phi_next, J_next])
```

### Seção de poincaré

```python
# wf/poincare.py
import numpy as np

def unwrap(a): return (a + np.pi)%(2*np.pi) - np.pi

def collect_section(step_fn, state0, dt, steps, phi1_star=0.0, burn=200, p1_from=None):
    s = np.array(state0, float)
    φ1_prev = unwrap(s[0]); star = unwrap(phi1_star)
    pts = []; cross = 0
    for _ in range(steps):
        s_prev = s.copy()
        s = step_fn(s, dt)
        φ1 = unwrap(s[0])
        if (φ1_prev < star <= φ1) and (p1_from is None or p1_from(s) > 0):
            # interp linear
            t = (star - φ1_prev) / (φ1 - φ1_prev + 1e-15)
            s_star = s_prev + t*(s - s_prev)
            cross += 1
            if cross > burn:
                pts.append(s_star.copy())
        φ1_prev = φ1
    return np.array(pts)
```

### Conversões e condição inicial

```python
# wf/coords.py
import numpy as np
def aa_to_cart(J, phi):
    J = np.maximum(J, 0.0)
    q = np.sqrt(2*J)*np.cos(phi)
    p = -np.sqrt(2*J)*np.sin(phi)
    return q,p

def cart_to_aa(q,p):
    J = 0.5*(q*q + p*p)
    phi = np.arctan2(-p, q)
    return J,phi
```

```python
# wf/init_cond.py
from scipy.optimize import newton

def find_J1_for_energy(E, J2, phi1, phi2, params, guess=0.1):
    def f(J1):
        if J1 < 0: return 1e9
        return H(phi1,phi2,J1,J2,params) - E
    try:
        x = newton(f, guess, tol=1e-10, maxiter=60)
        return x if x >= 0 else None
    except Exception:
        return None
```

### *Exercício 2.4 (Computacional)
**Construa uma seção de Poincaré no plano $p_2 \times q_2$ para $q_1=0$, $p_1 \ge 0$, $\alpha=0.1$, $\beta=0$ e energia total $E=0.18$. Observe que as condições $q_1=0, p_1>0$ equivalem à identidade $\phi_1=3\pi/2$. Indique os pontos fixos elípticos e hiperbólicos em seu gráfico. Anexe a figura obtida ao seu trabalho.**

---

A seguir, implementamos o laço principal para gerar a seção de Poincaré solicitada. O procedimento é:
1.  Inicializar o integrador com os parâmetros $\alpha=0.1$ e $\beta=0$.
2.  Definir uma grade de condições iniciais para $(q_2, p_2)$ no plano da seção.
3.  Para cada condição inicial $(q_2, p_2)$:
    a. Converter para variáveis de ação-ângulo $(J_2, \phi_2)$.
    b. Usar a condição de energia $H=E$ e a condição da seção $\phi_1=3\pi/2$ para encontrar o valor correspondente de $J_1$ numericamente.
    c. Se um $J_1$ válido (não negativo) for encontrado, iniciar a integração a partir do estado $(\phi_1, \phi_2, J_1, J_2)$.
    d. Coletar os pontos de cruzamento da seção usando o método `integrate_and_poincare`.
4.  Plotar todos os pontos $(q_2, p_2)$ coletados no mesmo gráfico.
5.  Identificar visualmente e anotar os pontos fixos, ilhas de estabilidade (em torno dos pontos elípticos) e a separatriz (associada aos pontos hiperbólicos).

```python
def collect_poincare_events(integrator, initial_state, dt, steps,
                            phi1_star, burn):
    """
    Coleta cruzamentos exatos com φ1 = phi1_star e p1>0
    usando interpolação linear entre passos adjacentes.
    """
    s = np.array(initial_state, float)
    pts = []
    crosses = 0

    def unwrap(a):  # mantém a fase contínua em (-π, π]
        return (a + np.pi) % (2*np.pi) - np.pi

    φ1 = unwrap(s[0]); star = unwrap(phi1_star)

    for _ in range(steps):
        s_prev = s.copy()
        φ1_prev = unwrap(s_prev[0])

        s = integrator.symplectic_euler_step(s_prev, dt)
        φ1_now = unwrap(s[0])

        # orientação p1>0 no estado "agora"
        # (se tiver um helper p1_from(J1,φ1), use; aqui, como estamos em (J,φ), p1>0 ↔ sin(φ1)<0)
        p1_pos = (np.sin(φ1_now) < 0)

        # cruzamento de baixo para cima: φ1_prev < star <= φ1_now
        if (φ1_prev < star <= φ1_now) and p1_pos:
            # interpolação linear na fase para estimar o estado "na seção"
            t = (star - φ1_prev) / (φ1_now - φ1_prev + 1e-15)
            s_star = s_prev + t * (s - s_prev)
            crosses += 1
            if crosses > burn:
                pts.append(s_star)

        φ1 = φ1_now

    return np.array(pts)
```

```python
def numeric_jacobian(f, x, eps=1e-6):
    y = f(x)
    J = np.zeros((len(y), len(x)))
    for k in range(len(x)):
        dx = np.zeros_like(x); dx[k] = eps
        J[:,k] = (f(x+dx) - y) / eps
    return J

def classify_and_plot(points, integrator, E, ax):
    # points = lista de (phi2*, J2*)
    from numpy.linalg import eigvals
    def P_res(v):
        return poincare_residual_vec(v, integrator, E)  # deve existir

    def num_jac(f, x, eps=1e-6):
        y0 = f(x)
        J = np.zeros((2,2))
        for k in range(2):
            dx = np.zeros_like(x); dx[k] = eps
            J[:,k] = (f(x+dx) - y0)/eps
        return J

    ell, hyp = [], []
    for phi2_fp, J2_fp in points:
        if J2_fp < 5e-4:  # central
            ell.append((phi2_fp, J2_fp)); continue
        J = num_jac(lambda v: P_res(v), np.array([phi2_fp, J2_fp]))
        DP = J + np.eye(2)
        ev = eigvals(DP)
        if np.iscomplex(ev[0]) and np.allclose(np.abs(ev),1,atol=0.08):
            ell.append((phi2_fp, J2_fp))
        else:
            hyp.append((phi2_fp, J2_fp))

    handles = []
    if ell:
        q2e, p2e = action_angle_to_cartesian(np.array([p[1] for p in ell]),
                                             np.array([p[0] for p in ell]))
        handles.append(ax.plot(q2e, p2e, 'go', ms=7, label='Fixos Elípticos')[0])
    if hyp:
        q2h, p2h = action_angle_to_cartesian(np.array([p[1] for p in hyp]),
                                             np.array([p[0] for p in hyp]))
        handles.append(ax.plot(q2h, p2h, 'rx', ms=8, mew=2, label='Fixos Hiperbólicos')[0])

    return {'handles': handles, 'elliptic': len(ell), 'hyperbolic': len(hyp)}
```

```python
import numpy as np
from scipy.optimize import root

PHI1_STAR = 1.5*np.pi   # seção q1=0, p1>0
DT_MAP = 0.05           # passo para construir o mapa (pode ser 0.03–0.07)
MAX_STEPS = 150000

def wrap(a):  # (-π,π]
    return (a + np.pi) % (2*np.pi) - np.pi

def first_return_on_section(integrator, E, phi2, J2, phi1_star=PHI1_STAR,
                            dt=DT_MAP, max_steps=MAX_STEPS):
    """Um 'passo' do mapa de Poincaré: volta na mesma seção com p1>0."""
    J1 = find_initial_J1(E, J2, phi1_star, phi2, integrator)
    if J1 is None or J1 <= 0:
        return None  # fora da superfície de energia

    s = np.array([phi1_star - 1e-9, phi2, J1, J2], float)  # garante 'abaixo' da seção
    star = wrap(phi1_star)
    phi1_prev = wrap(s[0])

    for _ in range(max_steps):
        s_prev = s.copy()
        s = integrator.symplectic_euler_step(s, dt)
        phi1_now = wrap(s[0])

        crossed = (phi1_prev <= star) and (phi1_now > star)
        if crossed:
            t = (star - phi1_prev) / (phi1_now - phi1_prev + 1e-15)
            s_star = s_prev + t*(s - s_prev)
            # orientação p1>0
            J1s = max(0.0, s_star[2])
            p1s = -np.sqrt(2.0*J1s)*np.sin(s_star[0])
            if p1s > 0:
                return s_star  # [phi1*, phi2', J1', J2']
        phi1_prev = phi1_now
    return None

def poincare_residual(x, integrator, E):
    """r = P(x) - x  (x=[phi2, J2])"""
    phi2, J2 = x
    ret = first_return_on_section(integrator, E, phi2, J2)
    if ret is None:
        return np.array([1e3, 1e3])  # penaliza fora do domínio
    phi2p, J2p = wrap(ret[1]), ret[3]
    return np.array([wrap(phi2p - phi2), J2p - J2])

def find_fixed_points(E, integrator,
                      phi2_span=0.6, J2_min=0.04, J2_max=0.22,
                      n_phi=9, n_J=10, n_keep=10, tol_merge=1e-3):
    """varre grossamente, escolhe bons chutes e refina com Newton (root)."""
    seeds = []
    phi2_grid = np.linspace(PHI1_STAR - phi2_span, PHI1_STAR + phi2_span, n_phi)
    J2_grid   = np.linspace(J2_min, J2_max, n_J)

    # varredura grosseira
    for phi2 in phi2_grid:
        for J2 in J2_grid:
            r = poincare_residual([phi2, J2], integrator, E)
            rn = np.linalg.norm([wrap(r[0]), r[1]])
            if np.isfinite(rn):
                seeds.append((rn, phi2, J2))
    seeds.sort(key=lambda t: t[0])
    seeds = seeds[:n_keep]  # melhores candidatos

    # refino + deduplicação
    sols = []
    for _, phi2_0, J2_0 in seeds:
        sol = root(lambda x: poincare_residual(x, integrator, E),
                   x0=np.array([phi2_0, J2_0]), tol=1e-12)
        if sol.success:
            phi2_fp, J2_fp = wrap(sol.x[0]), sol.x[1]
            # mescla soluções próximas
            if all(np.hypot(wrap(phi2_fp-s[0]), J2_fp-s[1]) > tol_merge for s in sols):
                sols.append((phi2_fp, J2_fp))
    return sols  # lista de (phi2*, J2*)
```

```python
import numpy as np

def fixed_points_WF(E, alpha, phi1_star=1.5*np.pi):
    pts = []  # (q2, p2, kind, C, J2, J1)
    for C in (+1.0, -1.0):  # ramos cos(2Δ)=±1
        k = (5 - alpha*C)/(1 - alpha*C)
        a = -k**2 - 3*k + 1 + alpha*C*k
        b = k + 1
        # resolve a J2^2 + b J2 - E = 0
        disc = b*b + 4*a*E
        if disc < 0: 
            continue
        for s in (+1.0, -1.0):
            if abs(a) < 1e-14:  # caso linear raro
                J2 = E/b
            else:
                J2 = (-b + s*np.sqrt(disc))/(2*a)
            if not np.isfinite(J2) or J2 <= 0:
                continue
            J1 = k*J2
            if J1 < 0:
                continue
            # ângulos possíveis conforme C
            if C == +1.0:
                phi2_list = [phi1_star, phi1_star - np.pi]  # 3π/2 e π/2
            else:
                phi2_list = [phi1_star - np.pi/2, phi1_star - 3*np.pi/2]  # π e 0
            for phi2 in phi2_list:
                # normaliza ângulo
                phi2 = (phi2 + 2*np.pi) % (2*np.pi)
                q2 = np.sqrt(2*J2)*np.cos(phi2)
                p2 = -np.sqrt(2*J2)*np.sin(phi2)
                # tipo "esperado": C=+1 → verticais; C=-1 → laterais
                kind = 'lateral (C=-1)' if C < 0 else 'vertical (C=+1)'
                pts.append((q2, p2, kind, C, J2, J1))
    # sempre há o fixo central (J2=0 ⇒ q2=p2=0), elíptico em E baixos:
    pts.append((0.0, 0.0, 'central', None, 0.0, E))  # J1≈E quando J2→0
    return pts
```

```python
E_2_9 = 0.18
ALPHA_BETA_2_9 = 0.02
PHI1_SECTION = 1.5 * np.pi

integrator_2_9 = WalkerFordIntegrator(alpha=ALPHA_BETA_2_9, beta=ALPHA_BETA_2_9)

fig, ax = plt.subplots(figsize=(10, 8))

initial_conditions = [
    (0.1, 0.0), (0.2, 0.0), (0.3, 0.0), (0.4, 0.0), (0.5, 0.0), (0.6, 0.0),
    (0.7, 0.0), (0.8, 0.0), (0.85, 0.0), (0.45, 0.45), (0.6, 0.1)
]

for q2_0, p2_0 in initial_conditions:
    if q2_0 == 0 and p2_0 == 0: continue
    
    J2_0, phi2_0 = cartesian_to_action_angle(q2_0, p2_0)
    J1_0 = find_initial_J1(E_2_9, J2_0, PHI1_SECTION, phi2_0, integrator_2_9, J1_guess=0.05)

    if J1_0 is not None:
        initial_state = [PHI1_SECTION, phi2_0, J1_0, J2_0]
        poincare_data = integrator_2_9.integrate_and_poincare(initial_state, dt=0.2, num_steps=20000)
        
        if len(poincare_data) > 0:
            J2_p = poincare_data[:, 3]
            phi2_p = poincare_data[:, 1]
            q2_p, p2_p = action_angle_to_cartesian(J2_p, phi2_p)
            ax.plot(q2_p, p2_p, '.', ms=0.8)

ax.set_title(f'Seção de Poincaré para $E={E_2_9}, \alpha=\beta={ALPHA_BETA_2_9}$', fontsize=16)
ax.set_xlabel('$q_2$', fontsize=14)
ax.set_ylabel('$p_2$', fontsize=14)
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.set_aspect('equal', adjustable='box')
ax.grid(True, linestyle='--', alpha=0.6)

# Anotações das Ressonâncias
ax.annotate('Ressonância 2:2 (distorcida)', xy=(0.6, 0.5), xytext=(0.6, 0.8),
            arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, ha='center')
ax.annotate('Ressonância 2:3', xy=(-0.8, 0.0), xytext=(-0.8, -0.4),
            arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, ha='center')
ax.annotate('Mar Caótico', xy=(0.2, 0.7), xytext=(-0.4, 0.8),
            arrowprops=dict(facecolor='black', shrink=0.05), fontsize=12, ha='center')
plt.show()
```

Aumentando ainda mais a energia para $E=0.2$, a estrutura do espaço de fase se torna ainda mais complexa. As ressonâncias primárias podem se tornar instáveis, e em torno delas podem surgir **ressonâncias secundárias**. Estas são ilhas de estabilidade menores que aparecem em torno dos pontos fixos elípticos das ressonâncias primárias. A sua presença é um forte indicador de comportamento caótico complexo. Nesta seção, procuraremos por essas pequenas cadeias de ilhas que orbitam as ilhas maiores das ressonâncias primárias.
